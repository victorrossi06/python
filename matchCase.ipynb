```python
#Ex. 3

tipo = input("Digite a letra (A/B): ")
status = input("Digite o status: ")

match (tipo, status):
    case("A", "ativo"):
        print("Acesso!")
    case("B", "inativo"):
        print("Outro acesso!")
    case _:
        print("Combinação não reconhecida.")
```
