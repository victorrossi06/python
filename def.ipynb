```python
# Estrutura com recursão
def it(n):
    if n == 0:
        return 1
    return n * it ( n - 1 )

it()
```

```python
# Estrutura com repetição

def it(n):

    result = 1
    for i in range(1, n+1):
        result *= i

    return result
%time it(500)
```

```python
#EXERCICIO 1

aluno = {
    "nome": input("Digite seu nome: "),
    "nota1": int(input("Digite a primeira nota: ")),
    "nota2": int(input("Digite a segunda nota: ")),
    "nota3": int(input("Digite a terceira nota:" ))
}
media = ((aluno["nota1"])+(aluno["nota2"])+(aluno["nota3"]))/3

aluno["media"] = media

if media > 6:
    situacao = "Aprovado"
else:
    situacao = "Reprovado"

print(aluno)
print(situacao)
```

```python
#EXERCICIO 2

conta = {
    "titular": input("Digite seu nome: "),
    "saldo": 3400
}
print(conta)

saque = int(input("Digite o valor que quer sacar: "))

if saque > conta["saldo"]:
    print("Você não possui saldo suficiente para este saque!")

else:
    conta["saque"] = saque
    conta["saldo"] = conta["saldo"] - saque
    print("Saque aprovado!")

print(conta)
```

```python

#EXERCICIO 3

livro = {
    "paginas_lidas": float(input("Digite a quantidade de páginas lidas: ")),
    "total_paginas": 328
}

percentual_lido = (livro["paginas_lidas"] * 100) / livro["total_paginas"]
livro["percentual_lido"] = percentual_lido

if percentual_lido < 25:
    classificacao = 'Iniciante'
elif percentual_lido < 50:
    classificacao = 'Em progresso'
elif percentual_lido < 75:
    classificacao = 'Avançado'
elif percentual_lido < 100:
    classificacao = 'Finalizando...'
else:
    classificacao = 'Concluído'

livro["classificacao"] = classificacao
print(livro)

```

```python
#EXERCICIO 4

cliente = {
    "nome": input("Digite seu nome: "),
    "valor": int(input("Digite o valor da sua compra: ")),
    "desconto": input("Você tem desconto? s/n")
}

desconto = cliente.get("desconto", "s")
cliente["desconto"] = desconto
if desconto == "s":
    valor_descontado = cliente["valor"] * 0.25
    valor_oficial = cliente["valor"] - valor_descontado
else:
    valor_descontado = cliente["valor"] * 0.10
    valor_oficial = cliente["valor"] - valor_descontado

cliente["valor_oficial"] = valor_oficial
print(cliente)
    
```

```python
carrinho = {}
produto = ""

while produto != "sair":
    produto = input("Digite um produto: ")
    if produto == "sair":
        break
    preco = float(input("Preço: "))
    carrinho[produto] = preco

print(carrinho)
```
