```python
lista_nomes = []
lista_idades = []
lista_sexos = []

for i in range(2):
    print(f'\nDados da {i + 1}ª pessoa:')
    nome = input("Digite seu nome: ")
    lista_nomes.append(nome)
    idade = int(input("Digite sua idade: "))
    lista_idades.append(idade)
    sexo = input("Digite M para masculino e F para feminino: ")
    lista_sexos.append(sexo)

    print(lista_nomes) 
    print(lista_idades) 
    print(lista_sexos)
```

```python
lista_nomes = []
lista_idades = []
lista_sexos = []
contador = 0

while contador < 2:
    print(f'\nDados da {contador + 1}ª pessoa:')
    nome = input("Digite seu nome: ")
    lista_nomes.append(nome)
    idade = int(input("Digite sua idade: "))
    lista_idades.append(idade)
    sexo = input("Digite M para masculino e F para feminino: ")
    lista_sexos.append(sexo)
    contador += 1

print(lista_nomes) 
print(lista_idades) 
print(lista_sexos)
```
